# Imports

```python
import datetime
import papermill as pm
import os
```

# Main

```python
# update this path with the path of the 'scripts' folder downloaded from the repository
path_root_scripts_folder = "C:/path/to/consensus_case_study/scripts/"
```

```python
# Paths
path_to_notebook_case_study = path_root_scripts_folder + "single_fu_case_study_notebook.ipynb"
path_tmp_papermill_notebook = path_root_scripts_folder + "tmp-notebook-papermill.ipynb"
```

```python
start_general = datetime.datetime.now()
```

**Functional unit selection**

Select the functional units of interest.

```python
fu_list = [
    "L1.a",
    "L1.b",
    "L2.a.b",
    "L2.a.c",
    "L2.c.a",
    "L2.c.b",
]
```

**Convergence mode**

(De-)Activate the convergence mode for Monte Carlo simulations.
- To speed up the completion of the computations, set ``mc_convergence_=False``.\
This will provide you with a first quick glimpse of the package and results, limiting Monte Carlo simulations to 1000 runs.
- To generate results in convergence mode, set ``mc_convergence_=True``.\
This will run Monte Carlo simulations until convergence is reached (relative tolerance below 5\%, or maximum number of runs, e.g., default=500000)\
for a given statistical metric (e.g., default=mean). This hence requires more time to complete computations. 

```python
mc_convergence_=False
```

Run the notebook for each functional unit selected

```python
start_loc = datetime.datetime.now()
for curr_fu in fu_list:
    start = datetime.datetime.now()
    print(f"\nStart generation for fu={curr_fu}...")
    pm.execute_notebook(
        path_to_notebook_case_study, 
        path_tmp_papermill_notebook, 
        parameters=dict(fu_code_=curr_fu, 
                        mc_convergence_=mc_convergence_,
                        mc_n_runs_=1000, #only considered if mc_convergence_=False
                        sobol_active_=False,
                        subfigures_=True,
                        run_dynamic_ar6_b=False,
                        generate_gif_b=False),
    )
    end = datetime.datetime.now()
    print(f"\tfu={curr_fu}: run completed at {end.strftime('%Y-%m-%d %H:%M:%S')} (total duration = {str((end - start))})")

end_loc = datetime.datetime.now()
print(f"\nAll functional units completed at {end_loc.strftime('%Y-%m-%d %H:%M:%S')} (total duration = {str((end_loc - start_loc))})")
```

```python
# Remove the tmp notebook used by papermill
if os.path.exists(path_tmp_papermill_notebook):
    os.remove(path_tmp_papermill_notebook)
    print(f"Temporary file (papermill notebook) has been successfully deleted.")
```
